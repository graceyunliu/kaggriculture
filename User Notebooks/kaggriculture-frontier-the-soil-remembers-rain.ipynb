## Four games from the new rank one

The current leader's public games reveal a different economy: twelve workers by day ten, three quadrants, eight cows, six sheep, a forty-tile strawberry plateau, and a late wheat refill. Four public traces share the same farmer and market schedule; only a few hand actions differ near the finish.

The strongest trace is episode `89945750`, seat 1. It wins 633/640 across the established current, public, diverse, and historical archive. That result is replay evidence—not a public-score claim—but it is strong enough to become the new economic base.

```python
shape=pd.DataFrame({
    "day":[1,5,9,12,16,20,24,28,30],
    "hands":[0,3,10,12,12,12,12,12,12],
    "quadrants":[1,1,2,2,3,3,3,3,3],
    "cows":[3,4,5,8,8,8,8,8,8],
    "sheep":[1,2,6,6,6,6,6,6,6],
    "strawberries":[0,2,12,27,40,40,38,22,12],
    "melons":[7,10,11,6,10,9,6,2,0],
})
fig,(ax0,ax1)=plt.subplots(1,2,figsize=(15,5.4),gridspec_kw={"width_ratios":[.88,1.12]})
ax0.step(shape.day,shape.hands,where="post",lw=3,color=RAIN,label="hands")
ax0.step(shape.day,shape.quadrants*4,where="post",lw=3,color=GOLD,label="quadrants × 4")
ax0.plot(shape.day,shape.cows,lw=2.5,color=MOSS,marker="o",label="cows")
ax0.plot(shape.day,shape.sheep,lw=2.5,color=CLAY,marker="o",label="sheep")
ax0.set(title="Labor and herd arrive early",xlabel="day",ylabel="count")
ax0.legend(frameon=False,ncol=2); ax0.grid(); ax0.spines[["top","right"]].set_visible(False)
ax1.fill_between(shape.day,shape.strawberries,step="post",alpha=.55,color=RAIN,label="strawberry tiles")
ax1.step(shape.day,shape.melons,where="post",lw=3,color=GOLD,label="melon tiles")
ax1.axvspan(26,29,color=MOSS,alpha=.13,label="late wheat refill")
ax1.set(title="A broad strawberry plateau, then wheat",xlabel="day",ylabel="productive tiles")
ax1.legend(frameon=False); ax1.grid(); ax1.spines[["top","right"]].set_visible(False)
plt.tight_layout(); plt.show()
```

## A duel hiding inside the mirror

Subin's public trace supplies the field schedule; Savko's neighboring trace wins the economic duel. Replacing the whole policy throws away topology reliability, so this notebook separates the three action channels and asks which one actually moves the margin.

The result is unusually clean: the farmer and hand tapes contribute almost nothing. The market tape carries the frontier gain. Soil therefore keeps every weed repair and splices only Savko's public market actions.

```python
factorial=pd.DataFrame({
    "component":["Subin base","Savko farmer","Savko hands","Savko market","hands + market"],
    "mean margin":[0,0,-120.5,5636.5,6204.42],
    "wins":[6,6,6,9,10],
})
fig,(ax0,ax1)=plt.subplots(1,2,figsize=(15,5.3),gridspec_kw={"width_ratios":[1.12,.88]})
colors=["#718078","#718078",CLAY,RAIN,MOSS]
b=ax0.barh(factorial.component,factorial["mean margin"],color=colors,height=.58)
ax0.axvline(0,color="#dbe9de",lw=1,alpha=.5)
ax0.bar_label(b,labels=[f"{x:+,.0f}" for x in factorial["mean margin"]],padding=6,color="#eef5ea",weight="bold")
ax0.invert_yaxis(); ax0.set(title="Component factorial vs Subin",xlabel="mean reward margin")
ax0.grid(axis="x"); ax0.spines[["top","right","left"]].set_visible(False)
b=ax1.barh(factorial.component,factorial.wins,color=colors,height=.58)
ax1.bar_label(b,labels=[f"{x}/12" for x in factorial.wins],padding=6,color="#eef5ea",weight="bold")
ax1.set_xlim(0,11); ax1.invert_yaxis(); ax1.set(title="Matched wins",xlabel="wins / 12")
ax1.grid(axis="x"); ax1.spines[["top","right","left"]].set_visible(False)
plt.tight_layout(); plt.show()
```

## Keep the market edge; return the terminal ledger

Savko's market tape is strongest late, but letting it own the final liquidation widens tail risk. Five terminal cutoffs were tested. Step 680 retains the large anti-frontier gain while returning the last forty turns to Soil's already-validated terminal ledger.

On the focused screen, the parent wins only 3/12 against Savko and 9/12 against Subin. The step-680 composite rises to 6/12 and 12/12 respectively, without changing the worker schedule.

```python
cutoffs=pd.DataFrame({
    "policy":["Soil parent","680","696","704","716"],
    "vs Savko":[3,6,6,7,9],
    "vs Subin":[9,12,12,12,12],
    "vs J.Mubasshir":[11,11,11,10,10],
})
x=np.arange(len(cutoffs)); w=.23
fig,(ax0,ax1)=plt.subplots(1,2,figsize=(15,5.3),gridspec_kw={"width_ratios":[1.18,.82]})
for offset,column,color in [(-w,"vs Savko",RAIN),(0,"vs Subin",MOSS),(w,"vs J.Mubasshir",GOLD)]:
    bars=ax0.bar(x+offset,cutoffs[column],width=w,label=column,color=color)
    ax0.bar_label(bars,padding=3,color="#eef5ea",fontsize=9)
ax0.set_xticks(x,cutoffs.policy); ax0.set_ylim(0,13)
ax0.set(title="Terminal-cutoff screen",ylabel="wins / 12"); ax0.legend(frameon=False,ncol=3,loc="lower center")
ax0.grid(axis="y"); ax0.spines[["top","right"]].set_visible(False)
ax1.hlines(0,0,680,color="#4d6f5a",lw=8,alpha=.45)
ax1.hlines(0,680,720,color=MOSS,lw=8,alpha=.85)
ax1.scatter([0,680,720],[0,0,0],s=[70,150,70],color=[RAIN,GOLD,MOSS],zorder=3)
ax1.text(340,.12,"Savko market",ha="center",color="#eef5ea",weight="bold")
ax1.text(700,.12,"Soil ledger",ha="center",color="#eef5ea",weight="bold")
ax1.set_xlim(-12,732); ax1.set_ylim(-.28,.42); ax1.set_yticks([])
ax1.set(title="Selected market window",xlabel="environment step")
ax1.spines[["top","right","left"]].set_visible(False)
plt.tight_layout(); plt.show()
```

## The one regression explains its own repair

The raw composite lost one historical game by 600 coins. Chunk ablations showed that no single market block caused it; the effect accumulated across the market regime. Crucially, the scenario reveals its shop sequence before Savko and Subin first disagree at step 290.

If the first three unlocked shops are `PET_CAFE → FARMERS_MARKET → BRUNCH_SPOT`, the market channel stays with Subin. This observable-state branch restores the case to +36. It uses no opponent identity, hidden state, or seed number.

```python
events=pd.DataFrame({
    "step":[96,144,240,290,680],
    "event":["Pet Cafe","Farmers Market","Brunch Spot","first tape divergence","terminal ledger"],
    "kind":["observed","observed","observed","decision boundary","policy boundary"],
})
fig,(ax0,ax1)=plt.subplots(1,2,figsize=(15,5.3),gridspec_kw={"width_ratios":[1.2,.8]})
ax0.hlines(0,70,700,color="#4d6f5a",lw=4,alpha=.65)
palette={"observed":RAIN,"decision boundary":GOLD,"policy boundary":MOSS}
for i,row in events.iterrows():
    y=.15 if i%2==0 else -.17
    ax0.vlines(row.step,0,y,color=palette[row.kind],lw=2)
    ax0.scatter(row.step,0,s=115,color=palette[row.kind],edgecolor="#edf5e9",zorder=3)
    ax0.text(row.step,y,row.event,ha="center",va="bottom" if y>0 else "top",color="#eef5ea",fontsize=9,rotation=12)
ax0.axvspan(240,290,color=GOLD,alpha=.12)
ax0.set_xlim(60,710); ax0.set_ylim(-.48,.48); ax0.set_yticks([])
ax0.set(title="State is known before tapes diverge",xlabel="environment step")
ax0.spines[["top","right","left"]].set_visible(False)
repair=pd.DataFrame({"policy":["raw composite","state-routed composite"],"margin":[-600,36]})
b=ax1.barh(repair.policy,repair.margin,color=[CLAY,MOSS],height=.58)
ax1.axvline(0,color="#dbe9de",lw=1,alpha=.55)
ax1.text(-300,0,"−600",ha="center",va="center",color="#eef5ea",weight="bold")
ax1.text(18,1,"+36",ha="center",va="center",color="#102016",weight="bold",fontsize=9)
ax1.invert_yaxis(); ax1.set(title="Historical regression repaired",xlabel="reward margin")
ax1.grid(axis="x"); ax1.spines[["top","right","left"]].set_visible(False)
plt.tight_layout(); plt.show()
```

## Promotion gates

The frozen bytes pass 639/640 established archive games: 351/352 broad and 288/288 historical. More importantly, the market composite changes the frontier-facing tests: current rank-one mirrors rise from 44/64 to 64/64, and the untouched-seed holdout rises from 84/96 to 92/96.

Every run reaches `DONE`. These are local fixed-tape falsification tests—not a public-score estimate—and the single broad loss remains visible.

```python
gates=pd.DataFrame({
    "gate":["broad archive","historical seeds","rank-one mirrors","post-freeze blind"],
    "parent":[351,288,44,84],
    "Soil V11":[351,288,64,92],
    "games":[352,288,64,96],
})
y=np.arange(len(gates)); h=.31
fig,(ax0,ax1)=plt.subplots(1,2,figsize=(15,5.4),gridspec_kw={"width_ratios":[1.15,.85]})
p=ax0.barh(y+h/2,100*gates.parent/gates.games,height=h,color=RAIN,label="Soil V10")
c=ax0.barh(y-h/2,100*gates["Soil V11"]/gates.games,height=h,color=MOSS,label="Soil V11")
ax0.bar_label(p,labels=[f"{w}/{g}" for w,g in zip(gates.parent,gates.games)],padding=5,color="#eef5ea")
ax0.bar_label(c,labels=[f"{w}/{g}" for w,g in zip(gates["Soil V11"],gates.games)],padding=5,color="#eef5ea",weight="bold")
ax0.set_yticks(y,gates.gate); ax0.invert_yaxis(); ax0.set_xlim(40,105)
ax0.set(title="Frozen promotion gates",xlabel="win rate (%)"); ax0.legend(frameon=False)
ax0.grid(axis="x"); ax0.spines[["top","right","left"]].set_visible(False)
lift=pd.DataFrame({"gate":["rank-one","blind"],"wins gained":[20,8]})
b=ax1.barh(lift.gate,lift["wins gained"],color=[GOLD,RAIN],height=.58)
ax1.bar_label(b,labels=["+20 wins","+8 wins"],padding=7,color="#eef5ea",weight="bold")
ax1.invert_yaxis(); ax1.set_xlim(0,22); ax1.set(title="Frontier lift",xlabel="additional wins")
ax1.grid(axis="x"); ax1.spines[["top","right","left"]].set_visible(False)
plt.tight_layout(); plt.show()
```

## Live execution proof

Replay matrices explain fixed opponents. This cell tests the actual file interface: the frozen `main.py` plays a complete official-environment episode from each seat, and the notebook fails unless both agents finish `DONE`.


```python
from kaggle_environments import make

live_rows=[]
for side in (0,1):
    agents=[str(MAIN_PATH),"starter"] if side==0 else ["starter",str(MAIN_PATH)]
    env=make("kaggriculture",configuration={"seed":74001+side},debug=True)
    env.run(agents)
    statuses=[a.status for a in env.steps[-1]]
    rewards=[float(a.reward) for a in env.steps[-1]]
    assert statuses==["DONE","DONE"],statuses
    live_rows.append({"candidate seat":side,"turns":len(env.steps)-1,
                      "candidate reward":rewards[side],"starter reward":rewards[1-side],
                      "margin":rewards[side]-rewards[1-side],"statuses":" / ".join(statuses)})
live=pd.DataFrame(live_rows)
display(live.style.hide(axis="index").format({
    "candidate reward":"{:,.0f}","starter reward":"{:,.0f}","margin":"{:+,.0f}"})
        .background_gradient(subset=["margin"],cmap="YlGn"))
```

## Package integrity

The submission bundle must contain one root-level `main.py`. It is reopened immediately after creation, and the packed bytes must reproduce the frozen digest.


```python
BUNDLE_PATH=Path("submission.tar.gz")
with tarfile.open(BUNDLE_PATH,"w:gz") as bundle:
    bundle.add(MAIN_PATH,arcname="main.py")
with tarfile.open(BUNDLE_PATH,"r:gz") as bundle:
    members=bundle.getmembers()
    assert [m.name for m in members]==["main.py"]
    packed=bundle.extractfile(members[0]).read()
assert hashlib.sha256(packed).hexdigest()==EXPECTED_SHA256
display(HTML(f"<div style='padding:19px 23px;border:1px solid #53765e;border-radius:16px;background:linear-gradient(135deg,#172a20,#263b2e);color:#e5eee4'><b style='color:#63cfdf'>Bundle verified</b><br>one root file · {len(packed):,} bytes<br><code style='color:#e9c96b'>{EXPECTED_SHA256}</code></div>"))
```
